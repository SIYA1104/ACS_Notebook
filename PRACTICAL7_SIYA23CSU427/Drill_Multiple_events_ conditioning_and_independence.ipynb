<b>1)</b> A coin is tossed three times. What is the probability that there are exactly two heads, given that <p>


(a) the first outcome is a head? <p>

(b) the first outcome is a tail? <p>

(c) the first two outcomes are both heads? <p>

(d) the first two outcomes are both tails? <p>

(e) the first outcome is a head and the third outcome is a tail? <p>


```python
from itertools import product
sample_space = [''.join(outcome) for outcome in product('HT', repeat=3)]

event_two_heads = [s for s in sample_space if s.count('H') == 2]

def conditional_probability(condition_fn, event_fn):
    conditioned_space = [s for s in sample_space if condition_fn(s)]
    favorable = [s for s in conditioned_space if event_fn(s)]
    if len(conditioned_space) == 0:
        return 0
    return len(favorable) / len(conditioned_space)
conditions = {
    "(a) First outcome is a head": lambda s: s[0] == 'H',
    "(b) First outcome is a tail": lambda s: s[0] == 'T',
    "(c) First two outcomes are heads": lambda s: s[0:2] == 'HH',
    "(d) First two outcomes are tails": lambda s: s[0:2] == 'TT',
    "(e) First is head and third is tail": lambda s: s[0] == 'H' and s[2] == 'T'
}


event_condition = lambda s: s.count('H') == 2
for label, cond_fn in conditions.items():
    prob = conditional_probability(cond_fn, event_condition)
    print(f"{label}: Probability = {prob:.4f}")
```

<b>2)</b> A student must choose exactly two of the following three electives: art, French, or mathematics. The probability that he choses art is $\frac{5}{8}$ and the probability that he chooses both art and French is $\frac{1}{4}$. <p>



(a) What is the probability that he chooses mathematics? <p>

(b) What is the probability that he chooses either art or French? <p>


```python
P_AF = 0.3   # P(Art and French)
P_A = 0.6    # P(Art)

# Total probability across all 3 possible 2-subject combos is 1
# Possible combos: {A,F}, {A,M}, {F,M}

P_AM = P_A - P_AF                 
P_FM = 1 - (P_AF + P_AM)          # T

P_M = P_AM + P_FM
P_A_or_F = P_AF + P_AM + P_FM  

print(f"(a) Probability the student chooses Mathematics: {P_M:.2f}")
print(f"(b) Probability the student chooses either Art or French: {P_A_or_F:.2f}")
```

<b>8)</b> A coin is tossed three times. Consider the following five events? <p> 
<li> A: Heads on the first toss </li>
<li> B: Tails on the second toss </li>
<li> C: Heads on the third toss </li>
<li> D: All three outcomes the same </li>
<li> E: Heads on the first toss </li><p> 
(a) Which of the following pairs of events are independent? <p> 
(1) A, B <p> 

(2) A, D <p> 

(3) A, E <p> 

(4) D, E <p> 




##### Definitions:
Two events \( X \) and \( Y \) are independent if:
\[
P(X \cap Y) = P(X) \cdot P(Y)
\]

##### Probabilities:
- \( P(A) = P(B) = P(C) = P(E) = \frac{1}{2} \)
- \( P(D) = \frac{2}{8} = \frac{1}{4} \)
- \( P(A \cap B) = \frac{2}{8} = \frac{1}{4} \Rightarrow A \text{ and } B \text{ are independent} \)
- \( P(A \cap D) = \frac{1}{8} \Rightarrow A \text{ and } D \text{ are independent} \)
- \( P(A \cap E) = \frac{1}{2} \neq \frac{1}{4} \Rightarrow A \text{ and } E \text{ are not independent} \)
- \( P(D \cap E) = \frac{1}{8} = \frac{1}{4} \cdot \frac{1}{2} \Rightarrow D \text{ and } E \text{ are independent} \)

#### ✅ Final Answer:
- **(1) A and B**: ✅ Independent  
- **(2) A and D**: ✅ Independent  
- **(3) A and E**: ❌ Not independent  
- **(4) D and E**: ✅ Independent
