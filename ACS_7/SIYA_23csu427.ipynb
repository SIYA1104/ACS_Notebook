### LAWS OF PROBABILITY

#### Probability Laws Demonstrated in Python

```python
# Import necessary libraries
import numpy as np
def print_title(title):
    print(f"\n{'=' * 10} {title} {'=' * 10}")
```

```python
print_title("1. Basic Rule of Probability")
P_A = 0.7  # Let's say probability of event A is 0.7
print(f"P(A) = {P_A}")
print(f"Is P(A) between 0 and 1? {'Yes' if 0 <= P_A <= 1 else 'No'}")

print_title("2. Complementary Law")

P_A = 0.65
P_not_A = 1 - P_A
print(f"P(A) = {P_A}")
print(f"P(not A) = 1 - P(A) = {P_not_A}")

print_title("3a. Addition Law (Mutually Exclusive)")

P_A = 0.3
P_B = 0.4
P_A_or_B = P_A + P_B  # Since mutually exclusive
print(f"P(A) = {P_A}, P(B) = {P_B}")
print(f"P(A or B) = {P_A_or_B}")

print_title("3b. Addition Law (Not Mutually Exclusive)")

P_A = 0.5
P_B = 0.6
P_A_and_B = 0.2
P_A_or_B = P_A + P_B - P_A_and_B
print(f"P(A) = {P_A}, P(B) = {P_B}, P(A and B) = {P_A_and_B}")
print(f"P(A or B) = {P_A_or_B}")

print_title("4a. Multiplication Law (Independent)")

P_A = 0.5
P_B = 0.3
P_A_and_B = P_A * P_B
print(f"P(A) = {P_A}, P(B) = {P_B}")
print(f"P(A and B) = P(A) * P(B) = {P_A_and_B}")

print_title("4b. Multiplication Law (Dependent)")

P_A = 0.6
P_B_given_A = 0.5
P_A_and_B = P_A * P_B_given_A
print(f"P(A) = {P_A}, P(B|A) = {P_B_given_A}")
print(f"P(A and B) = P(A) * P(B|A) = {P_A_and_B}")

print_title("5. Law of Total Probability")

P_B1 = 0.3
P_B2 = 0.7
P_A_given_B1 = 0.2
P_A_given_B2 = 0.6

P_A = P_A_given_B1 * P_B1 + P_A_given_B2 * P_B2
print(f"P(A|B1) = {P_A_given_B1}, P(B1) = {P_B1}")
print(f"P(A|B2) = {P_A_given_B2}, P(B2) = {P_B2}")
print(f"P(A) = {P_A}")

print_title("6. Bayes' Theorem")

P_B = 0.3
P_A_given_B = 0.8
P_A = 0.5

P_B_given_A = (P_A_given_B * P_B) / P_A
print(f"P(B) = {P_B}, P(A|B) = {P_A_given_B}, P(A) = {P_A}")
print(f"P(B|A) = {P_B_given_A}")


```
