## SCENRIO:
#### You roll a fair 6-sided die, and you want to calculate the probability of each outcome, and the probability of rolling a number greater than 4.

```python
sample_space = [1, 2, 3, 4, 5, 6] # sample space for the die
total_outcomes = len(sample_space) # total outcomes
print("Probability of each face on a fair die:")
for outcome in sample_space:
    print(f"P({outcome}) = 1/{total_outcomes} = {1/total_outcomes}")
```

### Rolling a number greater than 4

```python
event = [x for x in sample_space if x > 4]
prob_event = len(event) / total_outcomes
print("\nEvent: Rolling a number greater than 4")
print(f"Favorable outcomes: {event}")
print(f"Probability = {len(event)}/{total_outcomes} = {prob_event}")
```

## Urn Problem Example:

### (Halmos) One hundred balls, 50 red and 50 green, are distributed between two urns. An urn is selected at random; and from it, a ball is drawn at random. Does the probability of getting a red ball this way depend on how the balls were distributed between the two urns?

```python
import random

urn1 = {'red': 1, 'green': 0}
urn2 = {'red': 49, 'green': 50}

assert urn1['red'] + urn2['red'] == 50
assert urn1['green'] + urn2['green'] == 50

def prob_red(urn1, urn2):
    total_urn1 = urn1['red'] + urn1['green']
    total_urn2 = urn2['red'] + urn2['green']
    
    p1 = (urn1['red'] / total_urn1) if total_urn1 > 0 else 0
    p2 = (urn2['red'] / total_urn2) if total_urn2 > 0 else 0

    prob = 0.5 * p1 + 0.5 * p2
    return prob

# probability of drawing a red ball
prob = prob_red(urn1, urn2)
print(f"Probability of drawing a red ball: {prob:.4f}")
```
